# Laboratorio 2: documentos flexibles con MongoDB

**Objetivo:** explorar documentos de observaciones con campos opcionales y etiquetas. Los ejemplos se ejecutan en memoria sin servidor MongoDB y muestran consultas equivalentes que se podrian realizar con PyMongo.

```python
from datetime import datetime, timezone

observations = [
    {
        "title": "Sesion de exploracion",
        "content": "Se encontraron valores faltantes en la columna edad.",
        "tags": ["calidad", "faltantes"],
        "course_id": 1,
        "created_at": datetime(2026, 9, 30, tzinfo=timezone.utc),
    },
    {
        "title": "Revision de etiquetas",
        "content": "Las categorias tienen escritura inconsistente.",
        "tags": ["limpieza"],
        "created_at": datetime(2026, 9, 29, tzinfo=timezone.utc),
        "source": {"file": "muestra_sintetica.csv", "row": 12},
    },
]

print(f"Documentos sinteticos cargados: {len(observations)}")
```

## Filtrar documentos

Una coleccion documental permite que un documento tenga `source` y otro no. La aplicacion debe decidir como validar campos comunes y opcionales.

```python
tag = "calidad"
matching = [item for item in observations if tag in item["tags"]]
[(item["title"], item.get("course_id"), item.get("source")) for item in matching]
```

## Consulta equivalente en PyMongo

Con MongoDB disponible, el filtro se veria asi. Se deja como ejemplo de referencia y no se ejecuta en esta celda para que el notebook sea autocontenido.

```python
# En una base MongoDB real:
# collection.find({"tags": "calidad"}).sort("created_at", -1)
# collection.insert_one({"title": "...", "content": "...", "tags": ["..."]})

assert all(isinstance(item["tags"], list) for item in observations)
assert any("source" not in item for item in observations)
print("Validacion: hay etiquetas en todos los documentos y el campo source es opcional.")
```
